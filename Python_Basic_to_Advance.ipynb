```python
#magic method
#how to print fractions

class Fraction:

  def __init__(self, x, y):
    self.num = x
    self.den = y

  def __str__(self):
    return'{}/{}'.format(self.num,self.den)

  def __add__(self, other):
    new_num = self.num * other.den + other.num * self.den
    new_den = self.den * other.den

    return'{}/{}'.format(new_num, new_den)

  def __sub__(self, other):
    new_num = self.num * other.den - other.num * self.den
    new_den = self.den * other.den

    return'{}/{}'.format(new_num, new_den)

  def __mul__(self, other):
    new_num = self.num*other.num
    new_den = self.den * other.den

    return'{}/{}'.format(new_num, new_den)

  def __trudiv__(self, other):
    new_num = self.num*other.den
    new_den = self.den * other.num

    return'{}/{}'.format(new_num, new_den)

  def convert_to_decimal(self):
    return self.num/self.den



```

```python
class Point:
  def __init__(self,x,y):
   self.x_cod = x
   self.y_cod = y

  def __str__(self):
   return'<{},{}>'.format(self.x_cod,self.y_cod)


  def euclidean_distance(self,other):
    return ((self.x_cod-other.x_cod)**2 + (self.y_cod-other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return (self.x_cod**2 + self.y_cod**2)**0.5

class Line:
  def __init__(self, A, B, C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A, self.B, self.C)

  def point_on_line(line, Point):
    if line.A*Point.x_cod + line.B*Point.y_cod + line.C == 0:
      return "lies on the line"
    else:
      return "does not lie on the line"

  def shortest_distance(line, Point):
    return abs(line.A*Point.x_cod + line.B*Point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5


```

```python
#Super Keyword

class Phone:

  def __init__(self, brand, price, camera):
    print('inside phone constructor')

    self.brand = brand
    self.price = price
    self.camera = camera


  def buy(self):
    print('buying a phone')


class Smartphone(Phone):

  def __init__(self, os, ram):
    self.os = os
    self.ram = ram
    print("inside smartphone constructor")


  def buy(self):
    print('buying a smartphone')
    #syntax to call parent buy method
    super().buy()

s = Smartphone("Android", 6) # Correct: create an instance of Smartphone
s.buy()
```

# Inhetritance in summary

1. a class can inherit from another class.

2. It improves code reusability.

3. constructor, method, attributes get inherited to the child class.

4. The parent class has no acess to child class.

5. Private properties of parent can not accessed by the child class.

6. super () is used to call parent class method and constructor.

7. child class can override attributes and methods, this is called method overriding.

```python
# raise exception

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdrwa(self,amount):
    if amount < 0:
      raise Exception('amount canot be -ve')
    if self.balance < amount:
      raise Exception('paisa nhi milega')
    self.balance = self.balance - amount

obj = Bank (15000)
try:
  obj.withdrwa(2000)
except Exception as e:
  print(e)
else:
  print(obj.balance)



```

```python
# Creating custom exception
class SecurityError(Exception):

    def __init__(self,message):
      print('message')

    def logout(self):
      print('logout')

class Google:

    def __init__(self,name, email, password, device):
      self.name = name
      self.email = email
      self.password = password
      self.device = device

    def login(self, email, password, device):
      if device != self.device:
        raise SecurityError('you are using wrong device')
      if email == self.email and password == self.password:
        print('welcome back sir')
      else:
        print('Credentials not correct')

obj = Google('Amit','amit.workyadav@gmail.com','1,2,3,4','android')

try:
  obj.login('amit.workyadav@gmail.com','1,2,3,4','Windows')
except SecurityError as e:
  print(e)
else:
    print(obj.name)
finally:
  print('database connection closed')

```

Parametric Density Estimation

```python
import numpy.random as normal
import numpy as np
import matplotlib.pyplot as plt
```

```python
sample = normal.normal(loc=50, scale=5, size=100)
```

```python
sample.mean()
```

```python
plt.hist(sample, bins =10)
```

```python
sample_mean = sample.mean()
sample_std = sample.std()
```

```python
from scipy.stats import norm
dist = norm(sample_mean, sample_std)
```

```python
values = np.linspace(sample.min(), sample.max(), 100)
```

```python
prob = [dist.pdf(value)for value in values]
```

```python
plt.hist(sample, bins =10, density = True)
plt.plot(values,prob)
```
